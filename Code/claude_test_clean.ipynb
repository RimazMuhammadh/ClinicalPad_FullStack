```python
# 2. Configuration - Load API key and set working model
load_dotenv()

# Get API key from .env file
BEDROCK_API_KEY = os.getenv("rimaz_bayseian-at-734154191535")
if BEDROCK_API_KEY:
    BEDROCK_API_KEY = BEDROCK_API_KEY.strip('"')

# Working configuration - Claude 3 Sonnet
REGION = "eu-west-2"  # London
MODEL_ID = "anthropic.claude-3-sonnet-20240229-v1:0"  # This model works!
URL = f"https://bedrock-runtime.{REGION}.amazonaws.com/model/{MODEL_ID}/invoke"

print("✅ Configuration loaded!")
print(f"Region: {REGION}")
print(f"Model: Claude 3 Sonnet")
print(f"API Key: {'✅ Found' if BEDROCK_API_KEY else '❌ Missing'}")
```

```python
# 3. Function to ask Claude
def ask_claude(message):
    """Send message to Claude 3 Sonnet and get response"""
    
    headers = {
        "Content-Type": "application/json",
        "Authorization": f"Bearer {BEDROCK_API_KEY}"
    }
    
    payload = {
        "anthropic_version": "bedrock-2023-05-31",
        "max_tokens": 1500,
        "temperature": 0.1,  # Low temperature for consistent math answers
        "messages": [
            {"role": "user", "content": message}
        ]
    }
    
    try:
        response = requests.post(URL, headers=headers, json=payload, timeout=30)
        response.raise_for_status()
        
        result = response.json()
        return result['content'][0]['text']
        
    except requests.RequestException as e:
        return f"❌ Error: {str(e)}"

print("✅ Function ready!")
```

```python
# 4. Test - Simple Math Question
print("=== MATH TEST ===")

math_question = "What is 156 × 73? Please show your calculation step by step."

print(f"Question: {math_question}")
print("\nClaude's Answer:")
print("-" * 40)

response = ask_claude(math_question)
print(response)

print("\n" + "="*60)
```

```python
# 5. Test Different Math Problems
print("=== MULTIPLE MATH TESTS ===")

math_problems = [
    "What is 25 + 37?",
    "Calculate 144 ÷ 12",
    "What is 15² (15 squared)?",
    "Find the square root of 81"
]

for i, problem in enumerate(math_problems, 1):
    print(f"\n{i}. {problem}")
    answer = ask_claude(problem)
    print(f"   Answer: {answer}")
    print("-" * 50)
```
